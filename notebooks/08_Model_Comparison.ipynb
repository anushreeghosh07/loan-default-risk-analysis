```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

lr_results = pd.read_csv("../results/lr_results.csv")
knn_results = pd.read_csv("../results/knn_results.csv")
dt_results = pd.read_csv("../results/dt_results.csv")
rf_results = pd.read_csv("../results/rf_results.csv")
```

```python
comparison = pd.concat(
    [lr_results, knn_results, dt_results, rf_results],
    ignore_index=True
)

comparison = comparison.round(4)

comparison
```

```python
comparison.to_csv("../results/model_comparison.csv", index=False)

comparison
```

```python
comparison_plot = comparison.set_index("Model")

ax = comparison_plot.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Performance Comparison of Machine Learning Models")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.tight_layout()

plt.show()
```

```python
best_model = comparison.loc[[comparison["F1 Score"].idxmax()]]

print("Best Model Based on F1 Score:")

best_model
```

Four machine learning models—Logistic Regression, K-Nearest Neighbors (KNN), Decision Tree, and Random Forest—were developed and evaluated for predicting loan default risk. Since the dataset is imbalanced, the F1-score was selected as the primary evaluation metric. Among all the models, Logistic Regression achieved the highest F1-score while maintaining competitive accuracy and ROC-AUC. Therefore, Logistic Regression was selected as the final model for deployment.
